# QuantumCanvas Visualization
This notebook implements a `QuantumCanvas` to visualize quantum mechanical wave functions and their probability densities.

```python
import numpy as np
import matplotlib.pyplot as plt

class QuantumCanvas:
    def __init__(self, x_range=(-5, 5), points=1000):
        self.x = np.linspace(x_range[0], x_range[1], points)
        self.fig, self.ax = plt.subplots(figsize=(10, 6))

    def plot_wavefunction(self, psi, label='Wave Function'):
        psi_val = psi(self.x)
        self.ax.plot(self.x, psi_val.real, label=f'Re({label})')
        if np.iscomplexobj(psi_val):
            self.ax.plot(self.x, psi_val.imag, '--', label=f'Im({label})')

        self.ax.set_title("Quantum Wave Function Visualization")
        self.ax.set_xlabel("Position (x)")
        self.ax.set_ylabel("Amplitude")
        self.ax.legend()
        self.ax.grid(True, alpha=0.3)

    def plot_probability_density(self, psi, label='Prob Density'):
        psi_val = psi(self.x)
        prob_density = np.abs(psi_val)**2
        self.ax.fill_between(self.x, prob_density, alpha=0.3, label=label)
        self.ax.plot(self.x, prob_density)
        self.ax.set_ylabel("Probability Density |ψ|²")
        self.ax.legend()
```

### Example: Harmonic Oscillator Ground State
Let's visualize the ground state of a Quantum Harmonic Oscillator.

```python
def harmonic_oscillator_ground(x):
    return np.exp(-x**2 / 2) * (np.pi)**(-0.25)

canvas = QuantumCanvas(x_range=(-4, 4))
canvas.plot_wavefunction(harmonic_oscillator_ground, label='ψ₀')
canvas.plot_probability_density(harmonic_oscillator_ground, label='|ψ₀|²')
plt.show()
```

### Example: Moving Gaussian Wave Packet
This demonstrates a wave function with a complex phase: $\psi(x) = e^{-(x-x_0)^2 / (2\sigma^2)} e^{ikx}$.

```python
def gaussian_wave_packet(x, x0=0, sigma=0.5, k=5):
    envelope = np.exp(-(x - x0)**2 / (2 * sigma**2))
    phase = np.exp(1j * k * x)
    return envelope * phase

packet_canvas = QuantumCanvas(x_range=(-5, 5))
packet_canvas.plot_wavefunction(gaussian_wave_packet, label='\\psi_{packet}')
packet_canvas.plot_probability_density(gaussian_wave_packet, label='|\\psi|^2')
plt.show()
```
